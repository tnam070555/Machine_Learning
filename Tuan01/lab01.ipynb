## Bài 2. Tạo các Tensor thường dùng

**Yêu cầu:** Tạo Tensor ngẫu nhiên, Tensor toàn số 0, toàn số 1, ma trận đơn vị.
Nhận xét sự khác nhau giữa `torch.rand()` và `torch.randn()`.
Tạo thêm một Tensor 5×5 chứa toàn giá trị 10.

**Giải thích các hàm:**
- `torch.rand(3, 4)`: tạo Tensor kích thước 3×4 với các giá trị ngẫu nhiên lấy từ **phân phối đều (uniform)** trong khoảng [0, 1).
- `torch.randn(3, 4)`: tạo Tensor kích thước 3×4 với các giá trị ngẫu nhiên lấy từ **phân phối chuẩn (normal/Gaussian)**, trung bình = 0, độ lệch chuẩn = 1 → giá trị có thể âm hoặc lớn hơn 1.
- `torch.zeros(2, 3)`: tạo Tensor 2×3 toàn giá trị 0.
- `torch.ones(2, 3)`: tạo Tensor 2×3 toàn giá trị 1.
- `torch.eye(4)`: tạo ma trận đơn vị (identity matrix) kích thước 4×4 — đường chéo chính là 1, còn lại là 0.
- `torch.full((5, 5), 10)`: tạo Tensor 5×5 với mọi phần tử đều bằng 10 (dùng cho yêu cầu thêm).

```python
import torch

a = torch.rand(3, 4)     # phân phối đều [0, 1)
b = torch.randn(3, 4)    # phân phối chuẩn (mean=0, std=1)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)

print("a (rand) =\n", a)
print("b (randn) =\n", b)
print("c (zeros) =\n", c)
print("d (ones) =\n", d)
print("e (eye) =\n", e)

# Yêu cầu thêm: Tensor 5x5 toàn giá trị 10
f = torch.full((5, 5), 10)
print("f (full 10) =\n", f)
```

## Bài 4. Slicing và Indexing

**Yêu cầu:** Truy xuất một phần dữ liệu từ Tensor nhiều chiều.
Cho biết shape của `a[1]` và `a[1:, 2:4]`. Lấy hai phần tử cuối cùng trên chiều thứ ba.

**Giải thích:**
- `a` có kích thước `(3, 4, 5)` → gồm 3 "khối", mỗi khối là ma trận 4×5.
- `a[1]`: lấy khối thứ hai (chỉ số bắt đầu từ 0), kết quả là một ma trận 2 chiều.
- `a[1:, 2:4]`: 
  - `1:` trên chiều 0 → lấy từ khối 1 đến hết (bỏ khối 0).
  - `2:4` trên chiều 1 → lấy hàng 2 và hàng 3 (2 hàng) trong mỗi khối còn lại.
  - Chiều 2 (cột) giữ nguyên vì không chỉ định.
- Để lấy 2 phần tử cuối trên chiều thứ ba (chiều cuối, index 2), dùng slicing `[..., -2:]`, trong đó `...` giữ nguyên các chiều trước đó.

```python
import torch

a = torch.randn(3, 4, 5)
print("Original tensor:")
print(a)
print("Shape a:", a.shape)

print("\nTensor a[1]:")
print(a[1])
print("Shape a[1]:", a[1].shape)

print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])
print("Shape a[1:, 2:4]:", a[1:, 2:4].shape)

# Yêu cầu thêm: lấy 2 phần tử cuối trên chiều thứ ba (chiều cuối cùng)
print("\nHai phần tử cuối trên chiều thứ ba (a[..., -2:]):")
print(a[..., -2:])
print("Shape:", a[..., -2:].shape)
```

## Bài 5. Reshape và Flatten

**Yêu cầu:** Thay đổi hình dạng Tensor mà không làm thay đổi số phần tử.
Kiểm tra số phần tử trước/sau reshape. Thử reshape thành 2×6.

**Giải thích:**
- `a.ravel()`: "làm phẳng" Tensor thành một vector 1 chiều, giữ nguyên thứ tự phần tử (duyệt theo hàng).
- `a.reshape(3, 2, 2)`: thay đổi hình dạng Tensor thành 3 chiều, miễn là tổng số phần tử không đổi (3×2×2 = 12 = 3×4).
- Điều kiện bắt buộc khi reshape: **tích các chiều mới phải bằng tích các chiều cũ** (bằng đúng số phần tử `numel()`), nếu không sẽ báo lỗi.
- `a.reshape(2, 6)`: 2×6 = 12, hợp lệ vì bằng số phần tử ban đầu (3×4 = 12).

```python
import torch

a = torch.randn(3, 4)
print("Original:")
print(a)
print("Số phần tử ban đầu (numel):", a.numel())

print("\nFlatten:")
print(a.ravel())
print("Số phần tử sau flatten:", a.ravel().numel())

print("\nReshape to 3 x 2 x 2:")
r1 = a.reshape(3, 2, 2)
print(r1)
print("Số phần tử sau reshape(3,2,2):", r1.numel())

# Yêu cầu thêm: reshape thành 2 x 6
print("\nReshape to 2 x 6:")
r2 = a.reshape(2, 6)
print(r2)
print("Số phần tử sau reshape(2,6):", r2.numel())
```

## Bài 6. Phép toán và nhân ma trận

**Yêu cầu:** Thực hiện phép toán phần tử và phép nhân ma trận bằng toán tử `@`.
Phân biệt `a * b` với `a @ b.T`. Cho biết shape của `a @ b.T`.

**Giải thích:**
- `a + b`, `a / b`, `a ** 2`, `a * b`: đây là các **phép toán theo từng phần tử (element-wise)** — yêu cầu hai Tensor phải cùng shape (hoặc broadcast được), kết quả cũng có shape giống hai Tensor gốc.
- `a @ b.T`: đây là **phép nhân ma trận (matrix multiplication)** theo quy tắc đại số tuyến tính, không phải nhân từng phần tử. `b.T` là chuyển vị của `b` để shape khớp cho phép nhân ma trận: `(2,3) @ (3,2) → (2,2)`.

```python
import torch

a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("a:")
print(a)
print("b:")
print(b)

print("\na + b:")
print(a + b)
print("\na / b:")
print(a / b)
print("\na^2:")
print(a ** 2)

# Yêu cầu: phân biệt a * b với a @ b.T
print("\na * b (element-wise):")
print(a * b)
print("Shape a*b:", (a * b).shape)

print("\nMatrix multiplication a @ b.T:")
print(a @ b.T)
print("Shape a @ b.T:", (a @ b.T).shape)
```

## Bài 7. Các hàm thống kê trên Tensor

**Yêu cầu:** Tính trung bình, độ lệch chuẩn, tổng tích lũy theo từng cột.
Thay `dim=0` bằng `dim=1` và nhận xét sự khác nhau.

**Giải thích:**
- `dim` chỉ định **chiều mà phép tính được thực hiện dọc theo** (chiều đó sẽ bị "thu gọn" lại sau khi tính, trừ `cumsum` giữ nguyên shape).
- Với Tensor 2 chiều `(hàng, cột)`:
  - `dim=0`: tính toán **dọc theo các hàng** → tức là gộp theo từng **cột** (mỗi cột cho ra 1 giá trị).
  - `dim=1`: tính toán **dọc theo các cột** → tức là gộp theo từng **hàng** (mỗi hàng cho ra 1 giá trị).
- `torch.cumsum(a, dim=...)`: tính tổng tích lũy (cộng dồn) theo chiều chỉ định, không làm giảm shape.

```python
import torch

a = torch.randn(3, 4)
print(a)

print("\nMean dim=0:")
print(torch.mean(a, dim=0))
print("\nStandard deviation dim=0:")
print(torch.std(a, dim=0))
print("\nCumulative sum dim=0:")
print(torch.cumsum(a, dim=0))

# Yêu cầu: thay dim=0 bằng dim=1
print("\n=== dim=1 ===")
print("Mean dim=1:")
print(torch.mean(a, dim=1))
print("\nStandard deviation dim=1:")
print(torch.std(a, dim=1))
print("\nCumulative sum dim=1:")
print(torch.cumsum(a, dim=1))
```

## Bài 8. Autograd cơ bản

**Yêu cầu:** Tạo biến `requires_grad=True`, dùng `backward()` để tính đạo hàm tự động.
Tính thủ công đạo hàm của y = x² tại x = 3.6 và so sánh với `x.grad`.
Đổi hàm thành y = 3x² + 2x + 1 và kiểm tra gradient.

**Giải thích:**
- `requires_grad=True`: yêu cầu PyTorch theo dõi (track) mọi phép toán trên `x` để có thể tính đạo hàm sau này (xây dựng đồ thị tính toán - computation graph).
- `y.backward()`: thực hiện lan truyền ngược (backpropagation), tự động tính đạo hàm `dy/dx` và lưu vào `x.grad`.
- `grad_fn=<MulBackward0>`: cho biết `y` được tạo ra từ phép nhân (`x * x`), PyTorch dùng thông tin này để biết cách tính đạo hàm ngược.
- Đạo hàm thủ công của `y = x²` là `dy/dx = 2x`.
- Đạo hàm thủ công của `y = 3x² + 2x + 1` là `dy/dx = 6x + 2`.

```python
import torch

x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

# Yêu cầu thêm: đổi hàm thành y = 3x^2 + 2x + 1
print("\n=== Đổi hàm mới: y = 3x^2 + 2x + 1 ===")
x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3 * x2 ** 2 + 2 * x2 + 1
y2.backward()
print("x2 =", x2)
print("y2 =", y2)
print("x2.grad =", x2.grad)
print("Đạo hàm thủ công 6x+2 =", 6 * 3.6 + 2)
```

## Bài 9. Gradient Descent tìm hệ số đa thức

**Yêu cầu:** Sinh dữ liệu từ đa thức y = x² + 2x + 3, dùng NAdam và autograd để tìm lại 3 hệ số.
So sánh hệ số học được với [1, 2, 3]. Giải thích vai trò của `optimizer.zero_grad()`, `mse.backward()`, `optimizer.step()`.

**Giải thích:**
- `np.poly1d([1, 2, 3])`: tạo đa thức `1·x² + 2·x + 3`, dùng để sinh dữ liệu "nhãn" `Y` từ `X` ngẫu nhiên.
- `XX = [X², X, 1]`: xây ma trận đặc trưng để bài toán trở thành **hồi quy tuyến tính** theo 3 hệ số `w = [w0, w1, w2]`, vì `y = w0·x² + w1·x + w2 = XX @ w`.
- `w = torch.randn(3, 1, requires_grad=True)`: khởi tạo 3 hệ số ngẫu nhiên, cần tính gradient vì đây là tham số cần học.
- `torch.optim.NAdam([w], lr=0.01)`: bộ tối ưu NAdam (biến thể của Adam có thêm Nesterov momentum), cập nhật `w` để giảm hàm mất mát (MSE).
- Vòng lặp huấn luyện (1000 epoch) thực hiện lặp lại 3 bước cốt lõi: xóa gradient cũ → tính lại gradient mới qua backward → cập nhật trọng số.

```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()
    if (epoch + 1) % 200 == 0:
        print(f"Epoch {epoch+1}, MSE = {mse.item():.6f}")

print("Learned coefficients:")
print(w)
```

## Bài 10. Giải hệ phương trình bằng Autograd

**Yêu cầu:** Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa 4 phương trình:
A+B=9, C-D=1, A+C=8, B-D=2.
Kiểm tra nghiệm. Nhận xét vì sao bài toán có thể có nhiều nghiệm.

**Giải thích:**
- Ý tưởng: thay vì giải hệ phương trình tuyến tính bằng đại số, ta biến bài toán thành **bài toán tối ưu**: định nghĩa sai số `y1, y2, y3, y4` là hiệu giữa vế trái và vế phải của từng phương trình, sau đó tối thiểu hóa tổng bình phương sai số `sqerr = y1²+y2²+y3²+y4²`.
- Nếu tìm được `A, B, C, D` sao cho `sqerr ≈ 0`, nghĩa là cả 4 phương trình gần như được thỏa mãn đồng thời.
- `A, B, C, D` đều được khởi tạo ngẫu nhiên với `requires_grad=True`, dùng optimizer NAdam để cập nhật qua 2000 epoch, tương tự cách huấn luyện mô hình học máy.

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)

# Yêu cầu: kiểm tra nghiệm với 4 phương trình
print("\nKiểm tra:")
print("A+B =", (A+B).item(), "(mục tiêu 9)")
print("C-D =", (C-D).item(), "(mục tiêu 1)")
print("A+C =", (A+C).item(), "(mục tiêu 8)")
print("B-D =", (B-D).item(), "(mục tiêu 2)")
```
